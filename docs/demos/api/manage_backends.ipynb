# Backends

Depending on your system and your `pyclesperanto` installation, various backend will be available.

The backends are the low-level language used to access and run processing kernel on your Computational Units (GPU, CPU, etc.).
By default, `pyclesperanto` comes with `OpenCL` which is a widely compatible backend that runs on most system. It is also possible to use `CUDA` or `Metal` if you have installed it and is compatible with your system.
See the [pyclesperanto installation instructions](https://clesperanto-doc.readthedocs.io/en/latest/docs/pyclesperanto/install.html) for more details.


```python
import pyclesperanto as cle
```

Importing the package will automatically load the first available backend and device it can find.
If no backend were found, or that the backend that was loaded does not find any compatible devices, the package will raise an error.

Otherwise, we can check which backend is used.




```python
cle.get_backend_name()
```

We are not surprised to have `OpenCL` here, it is the default backend and the most compatible with all systems.
Bet let see if we have other backend available.

```python
cle.list_available_backends()
```

We have an other backend available, `Metal`, which is available only for MacOS system. Let's select it.

```python
cle.select_backend("Metal")
cle.info()
```

We sucessfully switch to metal, and when requesting information on the system we find the Apple Processing Unit. 
We can of course return to `OpenCL` if prefered.

```python
cle.select_backend("opencl")
cle.info()
```

We can see here that `OpenCL` can manage Apple devices due to its wide compatiblity.
It is interesting to see that `cle.info()`, depending on the backend, does not provide the same informations. This comes from the differences between the language.
Native backends, here `Metal`, should be more accurate.

Now, if we try to select a backend that is not available.

```python
try:
    cle.select_backend("cuda")
except Exception as e:
    print(f"Error occurred: {e}")
    print("\tCUDA backend is not available. Please ensure that you have the necessary CUDA drivers installed and that your GPU supports CUDA.")
    print(f"\tReverting to: {cle.get_backend_name()}.")
```

The error was expected, so we can managed it to select an other backend.

**WARNING:** Being at the based of all device selection, memory management, and processing, the backend should be defined at the **start** of the processes and **should not be modified** afterward until the end of the processing!
