# Matrix operations with pyclesperanto

This notebook demonstrates matrix operations on `pyclesperanto.Array`: matrix multiplication
(`cle.multiply_matrix`), the transpose requirement for compatible shapes, and element-wise
matrix arithmetic with another matrix or with a scalar.

```python
import numpy as np
import pyclesperanto as cle

cle.select_device()
```

## 1. Define two matrices

Push two 2D arrays (matrices) to the GPU.

```python
matrix1 = cle.push(np.asarray([[1, 2, 3, 4], [6, 7, 8, 9]], dtype=np.float32))
matrix2 = cle.push(np.asarray([[6, 7, 8], [9, 10, 11]], dtype=np.float32))

matrix1.shape, matrix2.shape
```

## 2. Matrix multiplication

`cle.multiply_matrix()` performs a real matrix product `(n, k) x (k, m) = (n, m)`: the number
of columns of the first matrix must equal the number of rows of the second.

Here `matrix1` is `(2, 4)` and `matrix2` is `(2, 3)`, so they are not directly compatible;
`matrix1` needs to be transposed to `(4, 2)` first.

```python
matrix1.T.shape, matrix2.shape
```

```python
result = cle.multiply_matrix(matrix1.T, matrix2)
result.shape, result
```

## 3. Element-wise matrix operations

Element-wise arithmetic between two matrices of the same shape uses the regular operators
(equivalent to `add_images`, `multiply_images`, ...), just like any other array.

```python
other = cle.push(np.random.random(matrix1.shape).astype(np.float32))

matrix1 + other, matrix1 * other
```

## 4. Matrix and scalar

Element-wise operations between a matrix and a scalar work the same way as for any array.

```python
matrix1 * 2, matrix1 / 2, matrix1 + 10
```

## 5. Matrix and vector

### Matrix-vector multiplication

`cle.multiply_matrix()` also works for a matrix times a vector, provided the vector is
shaped as a `(k, 1)` column matrix so the `(n, k) x (k, 1) = (n, 1)` shape rule is respected.

```python
vector = cle.push(np.asarray([[1], [1], [1], [1]], dtype=np.float32))
vector.shape
```

```python
matrix_vector = cle.multiply_matrix(matrix1, vector)
matrix_vector.shape, matrix_vector
```

### Element-wise matrix-vector operations (broadcasting)

A row vector (shape `(1, m)`) or a column vector (shape `(n, 1)`) broadcasts against a
matrix in element-wise operators, just like in NumPy.

```python
row_vector = cle.push(np.asarray([[1, 2, 3, 4]], dtype=np.float32))
matrix1 + row_vector, matrix1 * row_vector
```

```python
column_vector = cle.push(np.asarray([[1], [2]], dtype=np.float32))
matrix1 + column_vector, matrix1 * column_vector
```

**Important:** element-wise operations work on arrays of any dimensionality, while true
matrix multiplication with `cle.multiply_matrix()` is restricted to 2D matrices.
